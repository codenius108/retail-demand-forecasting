```python
import pandas as pd
import matplotlib.pyplot as plt

weather = pd.read_csv("../data/processed/weather_clean.csv")

weather.head()
```

```python
weather.info()
```

```python
for city in weather["city"].unique():

    city_data = weather[weather["city"] == city]

    plt.figure(figsize=(12, 5))

    plt.plot(
        city_data["date"],
        city_data["temperature"]
    )

    plt.title("Temperature Over Time - " + city)
    plt.xlabel("Date")
    plt.ylabel("Temperature (°C)")

    plt.show()
```
